```python
%reset -f
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
cmap=np.array([[120,18,134],[70,130,180],[0,118,14],[196,58,250],[220,248,164],[230,148,34],[205,62,78],[255,255,255],[120,120,120]])/255
```

```python
PermAccuracy = pd.read_csv(r'all_input_perm_accuracy.csv')
TrueAccuracy = pd.read_csv(r"D:\research\HCP_parcellation_Lsym-AVR\CamCan\KRR parcel size to age\evaluation\Pearson's r.csv")
TrueAccuracyMean = pd.DataFrame(TrueAccuracy.mean())
nets = ['VIS','SMN','DAN','VAN','LIM','FPN','DMN','GM','All']
NPerm = 1000
```

```python
Pvalue = []
for i in range(len(TrueAccuracyMean)):
    CurrP = (1+(PermAccuracy.iloc[:,i]>TrueAccuracyMean.iloc[i,0]).sum())/(1+NPerm)
    Pvalue.append(CurrP)
TrueAccuracyMean['Pvalue'] = Pvalue
```

```python
PermAccuracyLong = PermAccuracy.melt()
```

```python
plt.rcParams['font.sans-serif'][0]='Arial'
print(plt.rcParams['font.sans-serif'])
```

```python
sns.set_theme(style="white", rc={"axes.facecolor": (0, 0, 0, 0), 'axes.linewidth':1})
g = sns.FacetGrid(PermAccuracyLong, row="variable", aspect=9, height=1.2,hue='variable',palette=cmap)
g.figure.set_size_inches(3,2.6)
g.figure.set_dpi(300)
# Draw a density plot with a uniform color
g.map_dataframe(sns.kdeplot, x="value", fill=True, alpha=1)
g.map_dataframe(sns.kdeplot, x="value", color='black',linewidth=1)

ax = g.axes.flat[0]
LeftXlim = ax.get_xlim()
g.set(xlim=(LeftXlim[0],0.5))

for i,ax in enumerate(g.axes.flat):
    ax.text(0, 0.2, nets[i], color='black', fontsize=8,
                ha="left", va="center", transform=ax.transAxes)
    ax.axvline(TrueAccuracyMean[0][i],color='r',linewidth=2)
    ax.text(TrueAccuracyMean[0][i]+0.01,2,'P<0.001',color='black', fontsize=7,ha="left", va="center")

g.figure.subplots_adjust(hspace=.1)
g.set_titles("")
g.set(yticks=[], ylabel='')
g.set_xlabels("Accuracy of permutation",fontsize=10)
g.tick_params(axis='x',labelsize=8)
g.despine(left=True)
plt.subplots_adjust(left=.02,right=.92,bottom=.16)
plt.savefig('Ridgeplot.svg',format='svg')
```
