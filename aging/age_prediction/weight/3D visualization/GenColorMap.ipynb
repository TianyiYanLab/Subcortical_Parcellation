```python
%reset -f
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import pandas as pd
import os
```

```python
PfmPath = r'/path/to/your/data\CamCan\KRR parcel size to age\weights'
Nets = ['VIS','SMN','DAN','VAN','LIM','FPN','DMN']
for n in Nets:
    CurrWeights= pd.read_csv(f'{PfmPath}\{n}.csv')
    CurrWeightsMean = CurrWeights.mean()
    # 1. Set symmetrical range
    max_val = max(abs(CurrWeightsMean.min()), abs(CurrWeightsMean.max()))
    vmin, vmax = -max_val, max_val

    # 2. Create a normalization function
    norm = mcolors.Normalize(vmin=vmin, vmax=vmax)

    # 3. Choose a color map (RdBu: Red-White-Blue)
    cmap = plt.get_cmap('RdBu_r')  

    # 4. Get the color of each value
    colors = [cmap(norm(x))[:3] for x in CurrWeightsMean]

    # Create a mappable object for the colorbar
    sm = plt.cm.ScalarMappable(norm=norm, cmap=cmap)

    # Create a separate colorbar
    _,ax = plt.subplots(figsize=(8, 0.25),dpi = 100)  # Adjust size (width, height)
    cbar = plt.colorbar(sm, cax = ax,orientation='horizontal',ticks=[vmin,vmax])
    cbar.ax.tick_params(length = 10,width = 2,labelsize=20)
    if not os.path.exists(n):
        os.makedirs(n)

    plt.savefig(f'{n}\{n} colorbar.png', dpi=100, bbox_inches='tight')
    # Write to RGB file
    with open(f'{n}\{n} cmap.txt', 'w') as f:
        for r, g, b in colors:
            # Format each channel value
            line = f"{r:.3f} {g:.3f} {b:.3f}\n"
            f.write(line)
plt.savefig(f'colorbar.pdf',format='pdf')
```
