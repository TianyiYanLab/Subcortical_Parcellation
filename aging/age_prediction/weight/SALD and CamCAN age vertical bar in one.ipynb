```python
%reset -f
import pandas as pd
import scipy.io as sio
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
```

<font size=5>**先画Age**

```python
Nets = ['VIS','SMN','DAN','VAN','LIM','FPN','DMN']
Nucleus = ['ACC','AMY','CAU','HIP','GP','PUT','THA']
Ylabels = [f'{x}_{y}' for y in Nucleus for x in Nets]

# Assign colors for the bar plot
hue = pd.DataFrame(Ylabels)[0].str.split('_',expand=True)[1]
    # Get unique values and sort for consistency
unique_values = hue.unique()
num_unique = len(unique_values)

    # Generate colors (using the muted palette for high distinguishability)
colors = sns.color_palette("muted",n_colors=num_unique)

# Create color mapping dictionary
color_map = {value: colors[i] for i, value in enumerate(unique_values)}

# Add color column to DataFrame
hue = hue.to_frame(name='Nucleus')
hue['color'] = hue['Nucleus'].map(color_map)
```

```python
colorsnp = np.array(colors)
colorsnp = colorsnp*255
```

```python
def VertBar(data,hue,ax,Ylabels,direction):
    ax.spines['top'].set_visible(False)
    DataMean = np.mean(data,axis=1)
    SortIndex = np.argsort(DataMean)
    if direction=='pos':
        SortIndex = SortIndex[::-1]  # all with large absolute values are placed on top
    currHue = hue.iloc[SortIndex,:]
    currYlabels = np.array(Ylabels)[SortIndex]
    sns.barplot(data[SortIndex,:].T,errorbar='sd',ax = ax,orient="h")
    # Iterate over each bar and set colors manually
    for i, bar in enumerate(ax.patches):
        bar.set_color(currHue.iloc[i,1])
    if direction=='pos':
        ax.spines['right'].set_visible(False)
        ax.set_yticklabels(currYlabels,ha='right',va='center',fontsize= 7)
    else:
        ax.spines['left'].set_visible(False)
        ax.set_yticklabels(currYlabels,ha='left',va='center',fontsize= 7)
        ax.yaxis.tick_right()
        ax.yaxis.set_label_position('right')
    ax.xaxis.set_major_locator(MaxNLocator(integer=True))
    ax.tick_params(axis='x',pad=4,labelsize= 7)
    ax.tick_params(axis='both',width=1,length = 3)
    ax.spines['bottom'].set_linewidth(1)
    ax.spines['left'].set_linewidth(1)
```

```python
plt.rcParams['font.sans-serif'][0]='Arial'
fig,axes = plt.subplots(1,4,figsize=(6.9,8),dpi=300)

# CC pos
ax = axes[0]
currMat = sio.loadmat(r'/path/to/your/data\CamCAN\KRR parcel size to age\Weights\All nets PFM nuc by net.mat')
currMat = currMat['posMatNucByNet']
currMatR = currMat.reshape(49,100)
VertBar(currMatR,hue,ax,Ylabels,'pos')
ax.set_xlabel('CC positive',fontsize=8)
ax.axhline(6.5,color='red',linestyle='--')
ax.axhline(20.5,color='red',linestyle='--')

# SALD pos
ax = axes[1]
currMat = sio.loadmat(r'/path/to/your/data\SWU\KRR parcel size to age\weights\All nets PFM nuc by net.mat')
currMat = currMat['posMatNucByNet']
currMatR = currMat.reshape(49,100)
VertBar(currMatR,hue,ax,Ylabels,'pos')
ax.set_xlabel('SALD positive',fontsize=8)
ax.axhline(6.5,color='red',linestyle='--')
ax.axhline(18.5,color='red',linestyle='--')


# CC neg
ax = axes[2]
currMat = sio.loadmat(r'/path/to/your/data\CamCAN\KRR parcel size to age\Weights\All nets PFM nuc by net.mat')
currMat = currMat['negMatNucByNet']
currMatR = currMat.reshape(49,100)
VertBar(currMatR,hue,ax,Ylabels,'neg')
ax.set_xlabel('CC negative',fontsize=8)
ax.axhline(9.5,color='red',linestyle='--')
ax.axhline(20.5,color='red',linestyle='--')


# SALD neg
ax = axes[3]
currMat = sio.loadmat(r'/path/to/your/data\SWU\KRR parcel size to age\weights\All nets PFM nuc by net.mat')
currMat = currMat['negMatNucByNet']
currMatR = currMat.reshape(49,100)
VertBar(currMatR,hue,ax,Ylabels,'neg')
ax.set_xlabel('SALD negative',fontsize=8)
ax.axhline(6.5,color='red',linestyle='--')
ax.axhline(17.5,color='red',linestyle='--')

plt.tight_layout()
plt.subplots_adjust(left=.1, right=.9,bottom=.05,top=1,wspace=.8)
plt.savefig('figS2.svg',format='svg')
```
