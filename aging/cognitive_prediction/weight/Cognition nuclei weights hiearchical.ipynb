```python
%reset -f
import numpy as np
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from matplotlib import pyplot as plt
import scipy.io as sio
```

```python
Nets = ['VIS','SMN','DAN','VAN','LIM','FPN','DMN']
Nucleus = ['ACC','AMY','CAU','HIP','GP','PUT','THA']
Xlabels = [f'{x}_{y}' for y in Nucleus for x in Nets]
```

<font size=5>**正负权重一起平均**

```python
ct = 'Fluid intelligence'
currMat = sio.loadmat(rf'D:\research\HCP_parcellation_Lsym-AVR\CamCan\KRR parcel size to cognition\weights\results\{ct}\All nets PFM nuc by net.mat')
currMat = currMat['MatNucByNet']
currMatR = currMat.reshape(49,100)
currMatMean = np.mean(currMatR,axis=1)
currMatMean = currMatMean.reshape(-1,1)
# Z = linkage(currMatMean, 'average')
Z = linkage(currMatMean, 'complete')
f = fcluster(Z, 4, 'distance')
fig,ax = plt.subplots(figsize=(12, 9),dpi=300)
dn = dendrogram(Z,labels=Xlabels,leaf_rotation=0)
for label in ax.get_xticklabels():
    label.set_rotation(60)          # Rotate 45 degrees
    label.set_horizontalalignment('right')  # Align horizontally to the right
    label.set_verticalalignment('center')      # Align vertically to the top
    label.set_rotation_mode('anchor')
ax.set_title('Weights cluster of FI-predictive model',fontsize = 24)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
```
