```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
nucleus=['ACC','AMY','CAU','HIP','GP','PUT','THA']
currMat = sio.loadmat(rf'/path/to/your/data\CamCan\KRR parcel size to cognition\weights\results\Fluid intelligence\All nets PFM nuc by net.mat')
currMat = currMat['posMatNucByNet']
split_arrays = [currMat[i] for i in range(currMat.shape[0])]
for i in range(7):
    df = pd.DataFrame(split_arrays[i].T,columns=['VIS','SMN','DAN','VAN','LIM','FPN','DMN'])
    df=df.melt(var_name='Nets',value_name='Weights')
    model = ols('Weights ~ C(Nets)', data=df).fit()
    anova_results = sm.stats.AnovaRM(model, 'Nets')
    print(anova_results)
```
